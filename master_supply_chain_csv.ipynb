```python
import pandas as pd

# 1. Load the 9 raw CSV files directly from your local folder
shipping = pd.read_csv('Shipping.csv', encoding='latin1')
products = pd.read_csv('Products.csv', encoding='latin1')
payment_types = pd.read_csv('Payment Types.csv', encoding='latin1')
orders = pd.read_csv('Orders.csv', encoding='latin1')
order_items = pd.read_csv('Order Items.csv', encoding='latin1')
geography = pd.read_csv('Geography.csv', encoding='latin1')
departments = pd.read_csv('Departments.csv', encoding='latin1')
customers = pd.read_csv('Customers.csv', encoding='latin1')
categories = pd.read_csv('Categories.csv', encoding='latin1')

# 2. Handle missing values
products['Product Description'] = products['Product Description'].fillna(
    'No Description'
)
customers['Customer Zipcode'] = customers['Customer Zipcode'].fillna(0)
geography['Order Zipcode'] = geography['Order Zipcode'].fillna(0)
geography['Order State'] = geography['Order State'].fillna('Unknown')

# 3. Merge all datasets into one Master DataFrame
master = order_items.merge(orders, on='Order Id', how='left', suffixes=('', '_order'))
master = master.merge(shipping, on='Shipment_Id', how='left')
master = master.merge(payment_types, on='Payment_Type_ID', how='left')
master = master.merge(geography, on='Location_Id', how='left')
master = master.merge(
    customers,
    left_on='Order Customer Id',
    right_on='Customer Id',
    how='left',
    suffixes=('', '_cust'),
)
master = master.merge(products, on='Product Card Id', how='left')
master = master.merge(
    categories, left_on='Product Category Id', right_on='Category Id', how='left'
)
master = master.merge(departments, on='Department Id', how='left')

# 4. Save to CSV in your working directory
master.to_csv('master_supply_chain.csv', index=False)
print('master_supply_chain.csv successfully created and saved!')

# 5. Load and inspect
df = pd.read_csv('master_supply_chain.csv')
print('Head:')
print(df.head())
print('\nShape:', df.shape)
```
